# Introduction

The **IT Troubleshooting Assistant** is a Python-based project developed to help students identify and solve common computer problems. Students may experience issues related to network connections, hardware, software, and computer performance, but they may not always know how to fix them.

This system provides simple, step-by-step troubleshooting instructions, safety reminders, and user feedback options. Users can browse problems, search using keywords, or select a category to find suitable solutions. The main goal of this project is to make basic IT troubleshooting easier, faster, and more accessible to students while reducing their dependence on technical support for common issues.

## Section 2: Display Categories and Problems

These functions display all available categories and problems.
Users can also view problems belonging to a specific category.

```python
def show_categories():
    print("\nAVAILABLE CATEGORIES")
    categories = sorted({issue["category"] for issue in ISSUES.values()})
    for category in categories:
        print(f"- {category}")


def show_issues(category=None):
    print("\nAVAILABLE PROBLEMS")
    for number, issue in ISSUES.items():
        if category is None or issue["category"].lower() == category.lower():
            print(f'{number}. [{issue["category"]}] {issue["name"]}')
```

## Section 3: Search Function

This function searches for computer problems using keywords.
It checks the problem name, category, and symptoms, then returns
any matching problems.

```python
def search_issues(keyword):
    keyword = keyword.lower().strip()
    matches = []

    for number, issue in ISSUES.items():
        searchable_text = (
            issue["name"] + " " +
            issue["category"] + " " +
            " ".join(issue["symptoms"])
        ).lower()

        if keyword in searchable_text:
            matches.append((number, issue))

    return matches
```

## Section 4: Troubleshooting Guide and User Feedback

This function displays the selected problem and its recommended
solutions. It also shows a safety reminder and collects user
feedback on whether the solution helped.

```python
def give_solution(number):
    issue = ISSUES[number]

    print("\n" + "=" * 50)
    print("TROUBLESHOOTING GUIDE")
    print("=" * 50)
    print("Problem:", issue["name"])
    print("Category:", issue["category"])

    print("\nRecommended steps:")
    for index, step in enumerate(issue["steps"], start=1):
        print(f"{index}. {step}")

    print("\nSafety reminder:", issue["warning"])
    print("\nDid these steps solve your problem?")
    print("1. Yes")
    print("2. Partially")
    print("3. No")

    while True:
        feedback = input("Enter 1, 2 or 3: ").strip()

        if feedback in ("1", "2", "3"):
            break

        print("Invalid choice. Please enter 1, 2 or 3.")

    feedback_messages = {
        "1": "Glad to hear that! Your feedback is recorded for this session.",
        "2": "Thank you. Consider trying the remaining steps or contacting IT support.",
        "3": "Sorry the steps did not resolve the issue. Please contact IT support."
    }

    print(feedback_messages[feedback])
```

## Section 5: Main Menu

This function connects all the other functions. It allows users
to browse all problems, browse by category, search for a problem,
view solutions, and exit the application.

```python
def run_assistant():
    print("=" * 50)
    print("WELCOME TO THE IT TROUBLESHOOTING ASSISTANT")
    print("=" * 50)
    print("Get simple guidance for common computer problems.")

    while True:
        print("\nMAIN MENU")
        print("1. Browse all problems")
        print("2. Browse by category")
        print("3. Search for a problem")
        print("4. Exit")

        choice = input("Choose an option (1-4): ").strip()

        if choice == "1":
            show_issues()
            number = input(
                "\nEnter the problem number, or press Enter to return: "
            ).strip()

            if number in ISSUES:
                give_solution(number)
            elif number:
                print("Invalid problem number. Please try again.")

        elif choice == "2":
            show_categories()
            category = input(
                "\nEnter a category, or press Enter to return: "
            ).strip()

            if not category:
                continue

            valid_categories = {
                issue["category"].lower()
                for issue in ISSUES.values()
            }

            if category.lower() not in valid_categories:
                print("Category not found. Please choose a listed category.")
                continue

            show_issues(category)
            number = input(
                "\nEnter the problem number, or press Enter to return: "
            ).strip()

            if number in ISSUES and ISSUES[number]["category"].lower() == category.lower():
                give_solution(number)
            elif number:
                print("That problem does not belong to the selected category.")

        elif choice == "3":
            keyword = input(
                "Describe your problem using a keyword: "
            ).strip()

            if not keyword:
                print("Please enter a keyword.")
                continue

            matches = search_issues(keyword)

            if not matches:
                print("No matching problem found.")
                print("Try another keyword or contact IT support.")
                continue

            print("\nMATCHING PROBLEMS")
            for number, issue in matches:
                print(f'{number}. [{issue["category"]}] {issue["name"]}')

            number = input(
                "\nEnter a matching problem number, or press Enter to return: "
            ).strip()

            matching_numbers = [item[0] for item in matches]

            if number in matching_numbers:
                give_solution(number)
            elif number:
                print("Please select a number from the matching results.")

        elif choice == "4":
            print("\nThank you for using the IT Troubleshooting Assistant!")
            print("Goodbye!")
            break

        else:
            print("Invalid choice. Please enter a number from 1 to 4.")
```

## Section 6: Run the Application

This section starts the IT Troubleshooting Assistant and displays
the main menu for users.

```python
# Start the application
run_assistant()
```

## System Architecture

![Architecture Diagram](Architecture_Diagram_IT.jpeg)

IT Troubleshooting Assistant — System Architecture Specifications

1. Executive Summary
The IT Troubleshooting Assistant is a self-service diagnostic architecture designed to bridge the gap between student technical difficulties and overloaded IT support teams. By decoupling the presentation, processing, and data layers, the platform delivers real-time, deterministic, step-by-step resolution pathways for hardware, software, and networking issues.

2. Structural Layer Analysis
The system architecture of the IT Troubleshooting Assistant is organized into three distinct tiers, ensuring clear separation of concerns between the user interface, core processing logic, and the backend data repository.


    2.1. Presentation Layer (User Interface - UI)
    •	Scope: Encompasses all client-side components that interact directly with the user (Student). 
    •	Key Components:
    o	User (Student): The primary actor navigating the system to resolve technical issues. 
    o	Menu & Navigation Bar: Provides directional links (Home, Search, Categories, Troubleshooting). 
    o	Search Input: A text interface for entering keywords or specific error signatures. 
    o	Category Browser: Organizes troubleshooting entries into core technical domains: Hardware (HW), Software (SW), and Network (NW). 
    o	Solution Display: Renders step-by-step diagnostic workflows dynamically for the user.

   
    2.2 Application Layer (Core Processing Logic)
    •	Scope: Acts as the central middleware that intercepts user actions, executes matching logic, and formats responses. 
    •	Key Components:
    o	Query Processing & Matching Engine: The core processor (CPU representation) that manages incoming requests. 
    o	Search & Category Request Handlers: Processes user inputs from the UI to query the database. 
    o	Payload Formatter: Synthesizes matching problem data and troubleshooting steps into render-ready outputs for the presentation layer. 
    
    2.3 Data Layer (IT Troubleshooting Database)
    •	Scope: Manages persistence, storage, and telemetry feedback loops for all system knowledge. 
    •	Key Components:
    o	Knowledge Base: Stores known issues, error signatures, and solutions. 
    o	Category Data: Maintains taxonomic mapping for Hardware, Software, and Network groupings. 
    o	Guidance Data: Stores structured, sequential step-by-step instructions. 
    o	Troubleshooting Feedback Loop: Captures user validation telemetry to track resolution success or trigger manual IT escalation. 

4. Core Component Specifications:
   
    3.1 Presentation Layer (User Interface)
    •	Target Actor: End-user / Student experiencing technical friction. 
    •	Input Interface:
    o	Search Bar: Evaluates free-text inputs, keywords, and explicit error signatures (e.g., "DNS_PROBE_FINISHED_NXDOMAIN"). 
    o	Category Navigation: Provides structured taxonomy browsing divided into three root modules: 
    	⚙️ Hardware (HW): Power failures, display output anomalies, peripheral connectivity. 
    	💻 Software (SW): Operating system crashes, installer failures, runtime errors. 
    	🌐 Network (NW): Wi-Fi authentication, DHCP leasing failures, VPN drops. 
    •	Output Component: Displays step-by-step diagnostic workflows directly on the interface, complete with visual progress tracking. 


    3.2 Application Layer (Query Processing Engine)
    •	Role: Acts as the central middleware orchestrator. 
    •	Key Operations:
    1.	Parsing & Routing: Receives user action events from the UI via REST API endpoints. 
    2.	Matching Logic: Maps keyword parameters and category IDs against database indexes. 
    3.	Payload Formatting: Compiles retrieved step-by-step workflows into a clean JSON structure for render-ready display. 
    
    3.3 Data Layer (IT Troubleshooting Database)
    •	Schema Design: Structured knowledge base containing: 
    o	Category Matrix: Mapping for Hardware, Software, and Network domain hierarchies. 
    o	Issue Index: Catalog of known issue signatures and diagnostic metadata. 
    o	Resolution Workflows: Sequenced arrays of remediation steps tied to specific issue IDs. 


5.	Operational Data Flow & Feedback Protocol
Step	Initiator	Target Component	Protocol / Action	Data Exchanged
    a. Trigger	Student	Frontend (UI)	User Action (Input/Click)	Search string or Category selection 
    b. Request	Frontend (UI)	Query Engine	REST API Call (POST/GET)	Request JSON Payload 
    c. Query	Query Engine	Database	Database Query	Parametric Search Filters 
    d. Retrieval	Database	Query Engine	Dataset Response	Step-by-Step Resolution Record 
    e. Render	Query Engine	Solution Display	JSON Rendering	Sequential Guidance UI 
    f. Feedback Loop	Student	Database	User Rating Telemetry	Boolean Success / Escalation Event 

Closed-Loop Feedback Mechanism: Once a solution is displayed, the system prompts the user to verify resolution. Successful resolutions strengthen search weightings, while failed steps log an event for manual IT escalation.
